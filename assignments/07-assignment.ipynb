These trade-offs can reduce the value or accuracy of the outputs, meaning that resource constraints directly shape the way people interact with and benefit from LLMs on local hardware.

4. How does virtualisation contribute to the efficiency of cloud computing?


Virtualization is a core technology that allows cloud providers to create multiple, isolated environments on top of a single set of physical hardware, significantly boosting efficiency. By abstracting hardware resources into virtual machines (VMs) or containers, providers can run multiple operating systems and applications in parallel without interference. This consolidation greatly improves resource utilization and reduces operational overhead.

In addition, virtualization enables on-demand resource scaling by dynamically assigning CPU, memory, and storage based on changing workload requirements. This elasticity supports a pay-as-you-go model, helping organizations convert large capital expenditures into more manageable operational expenses. Large cloud providers also leverage economies of scale across thousands of customers, further reducing per-unit costs and allowing businesses to avoid guesswork around capacity planning.

Finally, virtualization simplifies deployment and portability by letting companies clone or migrate their environments across different data centers and regions worldwide. This flexibility lowers maintenance burdens and mitigates risks tied to hardware failures or geographic constraints. As a result, organizations can focus on delivering innovative features and services rather than devoting excessive time and effort to infrastructure management.

Example 1: AWS Lambda 

- Primary Uses: AWS Lambda is a serverless compute service that lets you run code in response to events (like an API call, an S3 file upload, or a scheduled timer) without managing servers. You just write your function code, set up triggers, and Lambda handles the scaling behind the scenes.
- Limitation: 
  - Cold Starts and Dependencies: If your function is rarely invoked or relies on heavy external libraries, it can experience a “cold start” delay. Packaging dependencies (e.g., large Python libraries) may also require extra configuration using Lambda layers.






Example 2: Amazon Comprehend

- Uses: Amazon Comprehend is a fully managed Natural Language Processing (NLP) service that enables users to extract valuable insights from unstructured text. It can analyze written content to determine sentiment (positive, negative, neutral, or mixed), identify key phrases, recognize named entities (such as people, places, and organizations), and detect the dominant language of a document. Comprehend is especially useful for businesses that need to analyze customer reviews, support tickets, social media posts, or survey responses at scale without building their own machine learning models.
  - One of the key benefits of Amazon Comprehend is its ease of use. Users can access its features through the AWS Management Console or API, and it doesn’t require any machine learning expertise. Real-time analysis tools built into the console allow users to quickly test the service by pasting sample text and viewing immediate results. It integrates well with other AWS services such as S3 and Lambda, enabling the automation of text analysis workflows for a wide range of applications.

- Limitations: Its support for non-English languages is relatively limited, with the most accurate results available for English and a few major global languages. It also lacks deep customization; users looking for domain-specific language models or custom entity recognition will need to turn to more flexible platforms like Amazon SageMaker. Additionally, Comprehend has usage limits under the free tier, and frequent or large-scale use can incur costs. Despite these constraints, it remains a powerful entry point for leveraging NLP capabilities in cloud-based applications.
