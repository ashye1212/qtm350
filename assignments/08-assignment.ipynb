1) List drivers ordered by `team` alphabetically and then by `victories` in descending order.

```python
query1 = """
SELECT * FROM drivers
ORDER BY team ASC, victories DESC;
"""
pd.read_sql_query(query1, conn)
```

2) List drivers who have won more races than the average number of victories across all drivers.

```python
query2 = """
SELECT * FROM drivers
WHERE victories > (
    SELECT AVG(victories) FROM drivers
);
"""
pd.read_sql_query(query2, conn)
```

3) Find all drivers with more than 50 podiums OR who drive for the 'Ferrari' team.

```python
query3 = """
SELECT * FROM drivers
WHERE podiums > 50 OR team = 'Ferrari';
"""
pd.read_sql_query(query3, conn)
```

4) Count the number of drivers for each `nationality`, but only include drivers who have more than 40 `podiums`. Order the result by count in descending order.

```python
query4 = """
SELECT nationality, COUNT(*) as driver_count
FROM drivers
WHERE podiums > 40
GROUP BY nationality
ORDER BY driver_count DESC;
"""
pd.read_sql_query(query4, conn)
```

5) Calculate the average `points` for each `team`, rounded to two decimal places.

```python
query5 = """
SELECT team, ROUND(AVG(points), 2) AS avg_points
FROM drivers
GROUP BY team;
"""
pd.read_sql_query(query5, conn)
```

6) List drivers who are in the top 20% based on their `points`.

```python
# First, get the 80th percentile value
points_df = pd.read_sql_query("SELECT points FROM drivers ORDER BY points DESC;", conn)
threshold = points_df['points'].quantile(0.80)
```

```python

query6 = f"""
SELECT * FROM drivers
WHERE points >= {threshold};
"""
pd.read_sql_query(query6, conn)
```

7) Calculate the total `victories`, total `podiums`, and average `points` for drivers with 'German' `nationality`.

```python
query7 = """
SELECT 
    SUM(victories) AS total_victories,
    SUM(podiums) AS total_podiums,
    ROUND(AVG(points), 2) AS avg_points
FROM drivers
WHERE nationality = 'German';
"""
pd.read_sql_query(query7, conn)
```

8) List all drivers whose `driver_name` contains both the letter 'a' and the letter 'e' (case-insensitive).

```python
query8 = """
SELECT * FROM drivers
WHERE LOWER(driver_name) LIKE '%a%'
  AND LOWER(driver_name) LIKE '%e%';
"""
pd.read_sql_query(query8, conn)
```

9) Retrieve the driver with the most `victories` within each `team`.

```python
query9 = """
SELECT d.*
FROM drivers d
INNER JOIN (
    SELECT team, MAX(victories) AS max_victories
    FROM drivers
    GROUP BY team
) tmax ON d.team = tmax.team AND d.victories = tmax.max_victories;
"""
pd.read_sql_query(query9, conn)
```

10) List drivers whose `points` are below the average `points` for their respective `team`.

```python
query10 = """
SELECT d.*
FROM drivers d
JOIN (
    SELECT team, AVG(points) AS avg_team_points
    FROM drivers
    GROUP BY team
) tavg ON d.team = tavg.team
WHERE d.points < tavg.avg_team_points;
"""
pd.read_sql_query(query10, conn)
```
